# Employee Attrition — EDA and Modeling
This notebook mirrors the modular training pipeline required by the project brief.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

df = pd.read_csv('../data/employee_data.csv')
df.head()
```

```python
df.info()
df.describe(include='all').T
```

```python
df['Attrition'].value_counts(normalize=True).mul(100).round(2)
```

```python
sns.set_theme(style='whitegrid')
rate = pd.crosstab(df['Department'], df['Attrition'], normalize='index')['Yes']
rate.mul(100).sort_values(ascending=False).plot(kind='bar', figsize=(9,5), title='Attrition Rate by Department', ylabel='Attrition rate (%)')
plt.tight_layout()
```

```python
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome')
plt.title('Monthly Income vs Attrition')
plt.tight_layout()
```

```python
numeric = df.select_dtypes('number').copy()
numeric['AttritionBinary'] = df['Attrition'].map({'No': 0, 'Yes': 1})
sns.heatmap(numeric.corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Heatmap Including Attrition')
plt.tight_layout()
```

```python
for col in ['OverTime','JobSatisfaction']:
    rate = pd.crosstab(df[col], df['Attrition'], normalize='index')['Yes']
    rate.mul(100).plot(kind='bar', title=f'Attrition Rate by {col}', ylabel='Attrition rate (%)', figsize=(7,5))
    plt.tight_layout()
    plt.show()
```
